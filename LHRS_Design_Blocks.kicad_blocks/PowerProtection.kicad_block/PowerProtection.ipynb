$V_{ovr} = \frac{R_{bot} * V_{ov}}{R_{top} + R_{bot}} $

```python
#Over Voltage Rising Threshold Calulations
Rbot = 7.5e3 #in Ohms
Rtop = 100e3 #in Ohms

Rtol = 0.01 #Tolerance of Resistors

Vovr_min = 1.165
Vovr_typ = 1.25
Vovr_max = 1.333

Vov_min = Vovr_min * ((Rbot*(1 + Rtol) + Rtop*(1 - Rtol)) / (Rbot*(1 + Rtol)))
Vov_typ = Vovr_typ * (Rbot + Rtop) / (Rbot)
Vov_max = Vovr_max * ((Rbot*(1 - Rtol) + Rtop*(1 + Rtol)) / (Rbot*(1 - Rtol)))
                      
print(f"Vov_min = {Vov_min}")
print(f"Vov_typ = {Vov_typ}")
print(f"Vov_max = {Vov_max}")
```

$ V_{un} = V_{uv}(1+ \frac{R_{bot}}{R_{top}}) $

```python
#Under Voltage Rising Threshold Calculations
Rbot = 16e3 #in Ohms
Rtop = 100e3 #in Ohms

Rtol = 0.01 #Tolerance of Resistors


Vuv_min = 1.16
Vuv_typ = 1.24
Vuv_max = 1.32

Vun_min = Vuv_min * (1 + (Rtop * (1 - Rtol)) / (Rbot * (1 + Rtol)))
Vun_typ = Vuv_typ * (1 + (Rtop/Rbot))
Vun_max = Vuv_max * (1 + (Rtop * (1 + Rtol) / (Rbot * (1 - Rtol))))

print(f" Vun_min = {Vun_min}")
print(f" Vun_typ = {Vun_typ}")
print(f" Vun_max = {Vun_max}")
```
